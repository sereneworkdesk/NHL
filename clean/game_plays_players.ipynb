Data Cleaning in Pandas

```python
import pandas as pd
```

```python
#understand the dataset

df_game_plays_players = pd.read_csv(r"C:\Users\zacle\Desktop\Serene\Project\NHL\ETL\extract\raw_data\game_plays_players.csv")
df_game_plays_players.head()
df_game_plays_players.info()
```

```python
# Create a new dataframe from df_game_plays_players (copying original to preserve data)
df_clean_game_plays_players = df_game_plays_players.copy()
```

Rename columns 

```python
#Rename columns for consistency

import re

# Function to add an underscore before uppercase letters and convert to lowercase
def rename_columns(col_name):
    return re.sub(r'([a-z0-9])([A-Z])', r'\1_\2', col_name).lower()

# Apply the function to all column names
df_clean_game_plays_players.columns = [rename_columns(col) for col in df_clean_game_plays_players.columns]
df_clean_game_plays_players.info()
```

Check : Null 

```python
# Null - All - check for missing or null values entire dataset
# Results : No null values

df_clean_game_plays_players.isnull().sum()
```

Check : Duplicates

```python
# Count unique 'play_id' values
unique_ids = df_clean_game_plays_players['play_id'].nunique()

# Count total number of rows
total_rows = len(df_clean_game_plays_players)

# Display the results
print(f"Unique game_ids: {unique_ids}")
print(f"Total rows: {total_rows}")
print(f"There are {total_rows - unique_ids} rows with duplicates.")

# Results: Unique game_ids: 3440757/ Total rows: 7586604/ There are 4145847 rows with duplicates.
```

```python
# Result : 1223800 duplicates
# Additional Duplicate Check for 'play_id', 'game_id', 'player_id', 'player_type' combinations
duplicates_id = df_clean_game_plays_players[df_clean_game_plays_players.duplicated(subset=['play_id', 'game_id', 'player_id', 'player_type'], keep=False)]

# Sort rows with duplicated results in ascending order by 'play_id', 'game_id', 'player_id', 'player_type'
duplicates_id_sorted = duplicates_id.sort_values(by=['play_id', 'game_id', 'player_id', 'player_type'], ascending=True)

# Count total number of rows with duplicates based on 'play_id', 'game_id', 'player_id', 'player_type'
total_duplicates_id = df_clean_game_plays_players.duplicated(subset=['play_id', 'game_id', 'player_id', 'player_type'], keep=False).sum()

# Count the unique combinations of 'play_id', 'game_id', 'player_id', 'player_type'
unique_combination = df_clean_game_plays_players[['play_id', 'game_id', 'player_id', 'player_type']].drop_duplicates().shape[0]

# Count total number of rows
total_rows = len(df_clean_game_plays_players)

# Display the results
print(f"Unique 'play_id', 'game_id', 'player_id', 'player_type' combinations: {unique_combination}")
print(f"Total rows: {total_rows}")
print(f"There are {total_rows - unique_combination} rows with duplicates based on 'play_id', 'game_id', 'player_id', 'player_type'.")

# Display the sorted duplicates for further review
print("Sorted Duplicates based on 'play_id', 'game_id', 'player_id', 'player_type':")
print(duplicates_id_sorted[['play_id', 'game_id', 'player_id', 'player_type']])
```
