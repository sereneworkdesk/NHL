```python
import pandas as pd
```

```python
#understand the dataset
# primaryPosition: The player's primary position on the ice:
# RW: Right Wing
# LW: Left Wing
# C: Center
# D: Defense
# G: Goalie

df_player_info = pd.read_csv(r"C:\Users\zacle\Desktop\Serene\Project\NHL\ETL\extract\raw_data\player_info.csv")
df_player_info.head()
df_player_info.info()
```

```python
# Create a new dataframe from df_game_goalie_stats (copying original to preserve data)
df_clean_player_info = df_player_info.copy()
```

Rename Columns

```python
#Rename columns for consistency 
#Function to add an underscore before uppercase letters and convert to lowercase
import re

def rename_columns(col_name):
    return re.sub(r'([a-z0-9])([A-Z])', r'\1_\2', col_name).lower()

# Apply the function to all column names
df_clean_player_info.columns = [rename_columns(col) for col in df_clean_player_info.columns]
df_clean_player_info.info()
```

Check Nulls

```python
# Null - All - check for missing or null values entire dataset
df_clean_player_info.isnull().sum()
```

```python
#Replace null values of the birth_state_province column with 'unknown'
df_clean_player_info['birth_state_province'] = df_clean_player_info['birth_state_province'].fillna('unknown')

#Removing null value rows will result in a Foreign Key error in game_goalie_stats. 
#Keep null values for the other columns as the number is trivial
#df_clean_player_info = df_clean_player_info.dropna()

df_clean_player_info['weight'] = df_clean_player_info['weight'].fillna(0)  # Fill the value with 0 

df_clean_player_info['height_cm'] = df_clean_player_info['height_cm'].fillna(0)  # Fill the value with 0

df_clean_player_info.isnull().sum()
```

```python
# duplicates : 0
# Count unique 'player_id' values
unique_ids = df_clean_player_info['player_id'].nunique()

# Count total number of rows
total_rows = len(df_clean_player_info)

# Display the results
print(f"Unique player_ids: {unique_ids}")
print(f"Total rows: {total_rows}")
print(f"There are {total_rows - unique_ids} rows with duplicates.")
```

```python
df_clean_player_info.info()
```

Change Data Type

```python
# Change data type syntax - df['column_name'] = df['column_name'].astype('desired_data_type')
df_clean_player_info['player_id'] = df_clean_player_info['player_id'].astype(str)
df_clean_player_info['first_name'] = df_clean_player_info['first_name'].astype(str)
```

```python
import numpy as np
# Ensure 'birth_date' is in datetime format, replacing invalid values with NaT
df_clean_player_info['birth_date'] = pd.to_datetime(df_clean_player_info['birth_date'], errors='coerce')

df_clean_player_info.info()
df_clean_player_info.head(3)
```

```python
print(df_clean_player_info.columns)
```

```python
#Convert weight from pounds to kilograms
df_clean_player_info['weight_kg'] = df_clean_player_info['weight'] * 0.453592

# Reorder columns to move 'weight_kg' to the 12th position
columns = list(df_clean_player_info.columns)
columns.insert(11, columns.pop(columns.index('weight_kg')))  # Move 'weight_kg' to position 12 (index 11)
df_clean_player_info = df_clean_player_info[columns]

# Print the DataFrame to verify
df_clean_player_info.head(5)
```

```python
# grp discussion - 22 Jan 
# to remove the redundant columns for simplifying data - height (duplicate in string format) 

columns_to_remove = ['height', 'weight']  
df_clean_player_info = df_clean_player_info.drop(columns=columns_to_remove)
df_clean_player_info.info()
```

```python
df_clean_player_info['first_name'] = df_clean_player_info['first_name'].astype(str)

max_length = df_clean_player_info['birth_state_province'].str.len().max()
print(max_length)  # Prints the maximum length of first names
```
