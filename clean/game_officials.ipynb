Data Cleaning in Pandas

```python
import pandas as pd
```

```python
#understand the dataset
df_game_officials = pd.read_csv(r"C:\Users\zacle\Desktop\Serene\Project\NHL\ETL\extract\raw_data\game_officials.csv")
df_game_officials.head()
df_game_officials.info()
```

```python
# Create a new dataframe from df_game_goalie_stats (copying original to preserve data)
df_clean_game_officials = df_game_officials.copy()
```

Check : Null 

```python
# Null - All - check for missing or null values entire dataset
df_clean_game_officials.isnull().sum()
```

```python
# Count unique 'game_id' values
unique_ids = df_clean_game_officials['game_id'].nunique()

# Count total number of rows
total_rows = len(df_clean_game_officials)

# Display the results
print(f"Unique game_ids: {unique_ids}")
print(f"Total rows: {total_rows}")
print(f"There are {total_rows - unique_ids} rows with duplicates.")
```

```python
#
# Additional Duplicate Check for 'game_id', 'official_name', 'official_type' combinations
duplicates_id = df_clean_game_officials[df_clean_game_officials.duplicated(subset=['game_id', 'official_name', 'official_type'], keep=False)]

# Sort rows with duplicated results in ascending order by 'game_id', 'official_name', 'official_type'
duplicates_id_sorted = duplicates_id.sort_values(by=['game_id', 'official_name', 'official_type'], ascending=True)

# Count total number of rows with duplicates based on 'game_id', 'official_name', 'official_type' 
total_duplicates = df_clean_game_officials.duplicated(subset=['game_id', 'official_name', 'official_type'], keep=False).sum()

# Count the unique combinations of 'game_id', 'official_name', 'official_type'
unique_game_official_combinations = df_clean_game_officials[['game_id', 'official_name', 'official_type']].drop_duplicates().shape[0]

# Count total number of rows
total_rows = len(df_clean_game_officials)

# Display the results
print(f"Unique 'game_id', 'official_name' and 'official_type' combinations: {unique_game_official_combinations}")
print(f"Total rows: {total_rows}")
print(f"There are {total_rows - unique_game_official_combinations} rows with duplicates based on 'game_id', 'official_name' and 'official_type'.")

# Display the sorted duplicates for further review
print("Sorted Duplicates based on 'game_id', 'official_name', 'official_type':")
print(duplicates_id_sorted[['game_id', 'official_name', 'official_type']])
```

```python
# Remove duplicates based on 'game_id', 'official_name' and 'official_type'
df_clean_game_officials = df_clean_game_officials.drop_duplicates(subset=['game_id', 'official_name','official_type'], keep='first')

# Display the cleaned dataframe (first 5 rows as an example)
df_clean_game_officials.head(5)
```

```python
# Change data type syntax - df['column_name'] = df['column_name'].astype('desired_data_type')
df_clean_game_officials['game_id'] = df_clean_game_officials['game_id'].astype(str)
df_clean_game_officials.info()
```

```python
#save file locally
df_clean_game_officials.to_csv(r"C:\Users\zacle\Desktop\Serene\Project\NHL\ETL\clean\game_officials.csv", index=False)
```
