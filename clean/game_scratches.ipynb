```python
import pandas as pd
```

```python
#understand the dataset

df_game_scratches = pd.read_csv(r"C:\Users\zacle\Desktop\Serene\Project\NHL\ETL\extract\raw_data\game_scratches.csv")
df_game_scratches.head(5)
df_game_scratches.info()
```

```python
# Create a new dataframe from df_game_plays_players (copying original to preserve data)
df_clean_game_scratches = df_game_scratches.copy()
```

Convert Datatype

```python
# Change data type syntax - df['column_name'] = df['column_name'].astype('desired_data_type')
df_clean_game_scratches['game_id'] = df_clean_game_scratches['game_id'].astype(str)
df_clean_game_scratches['team_id'] = df_clean_game_scratches['team_id'].astype(str)
df_clean_game_scratches['player_id'] = df_clean_game_scratches['player_id'].astype(str)
```

```python
df_clean_game_scratches.dtypes
```

```python
# Null - All - check for missing or null values entire dataset
# Results : No null values

df_clean_game_scratches.isnull().sum()
```

```python
# Result : 0
# Additional Duplicate Check for 'game_id', 'team_id', 'player_id' combinations
duplicates_id = df_clean_game_scratches[df_clean_game_scratches.duplicated(subset=['game_id', 'team_id', 'player_id'], keep=False)]

# Sort rows with duplicated results in ascending order by 'game_id', 'team_id', 'player_id'
duplicates_id_sorted = duplicates_id.sort_values(by=['game_id', 'team_id', 'player_id'], ascending=True)

# Count total number of rows with duplicates based on 'game_id', 'team_id', 'player_id'
total_duplicates_id = df_clean_game_scratches.duplicated(subset=['game_id', 'team_id', 'player_id'], keep=False).sum()

# Count the unique combinations of 'game_id', 'team_id', 'player_id'
unique_combination = df_clean_game_scratches[['game_id', 'team_id', 'player_id']].drop_duplicates().shape[0]

# Count total number of rows
total_rows = len(df_clean_game_scratches)

# Display the results
print(f"Unique 'game_id', 'team_id', 'player_id' combinations: {unique_combination}")
print(f"Total rows: {total_rows}")
print(f"There are {total_rows - unique_combination} rows with duplicates based on 'game_id', 'team_id', 'player_id'.")

# Display the sorted duplicates for further review
print("Sorted Duplicates based on 'game_id', 'team_id', 'player_id':")
print(duplicates_id_sorted[['game_id', 'team_id', 'player_id']])
```

```python
# Remove duplicates based on 'game_id', 'team_id', 'player_id'
df_clean_game_scratches = df_clean_game_scratches.drop_duplicates(subset=['game_id', 'team_id', 'player_id'], keep='first')

# Display the cleaned dataframe (first 5 rows as an example)
df_clean_game_scratches.head(5)

# No duplicates!
df_clean_game_scratches.isnull().sum()
```
