```python
import pandas as pd

df_team_info = pd.read_csv(r"C:\Users\zacle\Desktop\Serene\Project\NHL\ETL\extract\raw_data\team_info.csv")
df_team_info.head(10)
df_team_info.info()
```

```python
# Create a new dataframe from df_game_team_stats (copying original to preserve data)
df_clean_team_info = df_team_info.copy()
```

```python
# Null - All - check for missing or null values entire dataset
df_clean_team_info.isnull().sum()
```

```python
# duplicates : 28875 
# Count unique 'game_id' values
unique_ids = df_clean_team_info['team_id'].nunique()

# Count total number of rows
total_rows = len(df_clean_team_info)

# Display the results
print(f"Unique game_ids: {unique_ids}")
print(f"Total rows: {total_rows}")
print(f"There are {total_rows - unique_ids} rows with duplicates.")
```

```python
df_clean_team_info['team_id'] = df_clean_team_info['team_id'].astype(str)
df_clean_team_info.info()
```

```python
#Rename columns for consistency
#Function to add an underscore before uppercase letters and convert to lowercase
import re

def rename_columns(col_name):
    return re.sub(r'([a-z0-9])([A-Z])', r'\1_\2', col_name).lower()

# Apply the function to all column names
df_clean_team_info.columns = [rename_columns(col) for col in df_clean_team_info.columns]
df_clean_team_info.info()
```

```python
# grp discussion - 23 Jan - important on an enterprise level to leave all the extracted data intact
# as this is a project, we can have some leeway
# to remove the redundant columns for simplifying data - link has a url link (redundant to analysis)

# Drop 'link' column from df_clean_team_info
df_clean_team_info = df_clean_team_info.drop('link', axis=1)

# Verify by checking the updated DataFrame info
df_clean_team_info.info()
```
