```python
import pandas as pd

df_game_skater_stats = pd.read_csv(r"C:\Users\zacle\Desktop\Serene\Project\NHL\ETL\extract\raw_data\game_skater_stats.csv")
df_game_skater_stats.head()
df_game_skater_stats.info()
```

```python
# Create a new dataframe from df_game_skater_stats (copying original to preserve data)
df_clean_game_skater_stats = df_game_skater_stats.copy()
```

```python
#Rename columns for consistency

import re

# Function to add an underscore before uppercase letters and convert to lowercase
def rename_columns(col_name):
    return re.sub(r'([a-z0-9])([A-Z])', r'\1_\2', col_name).lower()

# Apply the function to all column names
df_clean_game_skater_stats.columns = [rename_columns(col) for col in df_clean_game_skater_stats.columns]

# Rename the columns 
df_clean_game_skater_stats = df_clean_game_skater_stats.rename(columns={'faceoff_taken': 'face_off_taken'})

df_clean_game_skater_stats.info()
```

```python
# Null - All - check for missing or null values entire dataset
df_clean_game_skater_stats.isnull().sum()
```

```python
# duplicates : 76841
# Count unique 'game_id' values
unique_ids = df_clean_game_skater_stats['game_id'].nunique()

# Count total number of rows
total_rows = len(df_clean_game_skater_stats)

# Display the results
print(f"Unique game_ids: {unique_ids}")
print(f"Total rows: {total_rows}")
print(f"There are {total_rows - unique_ids} rows with duplicates.")
```

```python
# 8738 duplicate rows
# Additional Duplicate Check for 'game_id', 'player_id' combinations
duplicates_id = df_clean_game_skater_stats[df_clean_game_skater_stats.duplicated(subset=['game_id', 'player_id'], keep=False)]

# Sort rows with duplicated results in ascending order by 'game_id', 'player_id'
duplicates_id_sorted = duplicates_id.sort_values(by=['game_id', 'player_id'], ascending=True)

# Count total number of rows with duplicates based on 'game_id', 'player_id'
total_duplicates = df_clean_game_skater_stats.duplicated(subset=['game_id', 'player_id'], keep=False).sum()

# Count the unique combinations of 'game_id', 'player_id'
unique_combinations = df_clean_game_skater_stats[['game_id', 'player_id']].drop_duplicates().shape[0]

# Count total number of rows
total_rows = len(df_clean_game_skater_stats)

# Display the results
print(f"Unique 'game_id', 'player_id' combinations: {unique_combinations}")
print(f"Total rows: {total_rows}")
print(f"There are {total_rows - unique_combinations} rows with duplicates based on 'game_id', 'player_id'.")

# Display the sorted duplicates for further review
print("Sorted Duplicates based on 'game_id', 'player_id':")
print(duplicates_id_sorted[['game_id', 'player_id']])
duplicates_id_sorted.head(10)
```

```python
# 8738 duplicate rows
# Additional Duplicate Check for 'game_id', 'player_id', 'team_id' combinations same as with 'game_id' and 'player_id'
duplicates_id = df_clean_game_skater_stats[df_clean_game_skater_stats.duplicated(subset=['game_id', 'player_id', 'team_id'], keep=False)]

# Sort rows with duplicated results in ascending order by 'game_id', 'player_id', 'team_id'
duplicates_id_sorted = duplicates_id.sort_values(by=['game_id', 'player_id', 'team_id'], ascending=True)

# Count total number of rows with duplicates based on 'game_id', 'player_id', 'team_id'
total_duplicates = df_clean_game_skater_stats.duplicated(subset=['game_id', 'player_id', 'team_id'], keep=False).sum()

# Count the unique combinations of 'game_id', 'player_id', 'team_id'
unique_combinations = df_clean_game_skater_stats[['game_id', 'player_id', 'team_id']].drop_duplicates().shape[0]

# Count total number of rows
total_rows = len(df_clean_game_skater_stats)

# Display the results
print(f"Unique 'game_id', 'player_id', 'team_id' combinations: {unique_combinations}")
print(f"Total rows: {total_rows}")
print(f"There are {total_rows - unique_combinations} rows with duplicates based on 'game_id', 'player_id', 'team_id'.")

# Display the sorted duplicates for further review
print("Sorted Duplicates based on 'game_id', 'player_id', 'team_id':")
print(duplicates_id_sorted[['game_id', 'player_id', 'team_id']])
duplicates_id_sorted.head(10)
```

```python
# Upon visual inspection of the 1st 10 rows shows that there are 8738 rows with true duplicates. 
# Remove duplicates based on 'game_id', 'player_id', 'team_id'(just to to be sure!)
df_clean_game_skater_stats = df_clean_game_skater_stats.drop_duplicates(subset=['game_id', 'player_id', 'team_id'], keep='first')

# Display the cleaned dataframe (first 30 rows as an example)
df_clean_game_skater_stats.head(10)
df_clean_game_skater_stats.info()
```

```python
# Duplicate Check : Duplicates resolved!
# Additional Duplicate Check for 'game_id', 'player_id' combinations
duplicates_id = df_clean_game_skater_stats[df_clean_game_skater_stats.duplicated(subset=['game_id', 'player_id'], keep=False)]

# Sort rows with duplicated results in ascending order by 'game_id', 'player_id'
duplicates_id_sorted = duplicates_id.sort_values(by=['game_id', 'player_id'], ascending=True)

# Count total number of rows with duplicates based on 'game_id', 'player_id'
total_duplicates = df_clean_game_skater_stats.duplicated(subset=['game_id', 'player_id'], keep=False).sum()

# Count the unique combinations of 'game_id', 'player_id'
unique_combinations = df_clean_game_skater_stats[['game_id', 'player_id']].drop_duplicates().shape[0]

# Count total number of rows
total_rows = len(df_clean_game_skater_stats)

# Display the results
print(f"Unique 'game_id', 'player_id' combinations: {unique_combinations}")
print(f"Total rows: {total_rows}")
print(f"There are {total_rows - unique_combinations} rows with duplicates based on 'game_id', 'player_id'.")

# Display the sorted duplicates for further review
print("Sorted Duplicates based on 'game_id', 'player_id':")
print(duplicates_id_sorted[['game_id', 'player_id']])
duplicates_id_sorted.head(30)
```

```python
# Change data type syntax - df['column_name'] = df['column_name'].astype('desired_data_type')
df_clean_game_skater_stats['game_id'] = df_clean_game_skater_stats['game_id'].astype(str)
df_clean_game_skater_stats['player_id'] = df_clean_game_skater_stats['player_id'].astype(str)
df_clean_game_skater_stats['team_id'] = df_clean_game_skater_stats['team_id'].astype(str)

df_clean_game_skater_stats.info()
```
