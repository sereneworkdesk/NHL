```python
#To change the directory containing the NHL .csv files.
import os
import shutil #shutil is shell utilities for moving files from directories

# Define the source directory and the new destination directory
source_path = r"C:\Users\zacle\.cache\kagglehub\datasets\martinellis\nhl-game-data\versions\2"
destination_path = r"C:\Users\zacle\Desktop\Serene\Project\NHL\ETL\extract\raw_data"

# Ensure the destination directory exists
if not os.path.exists(destination_path):
    os.makedirs(destination_path)  # Create the directory if it doesn't exist

# Get the list of files in the source directory
files = os.listdir(source_path)

# Loop through all files
for file in files:
    if file.endswith('.csv'):  # Only move CSV files
        # Generate the full path for the source and destination
        source_file_path = os.path.join(source_path, file)
        destination_file_path = os.path.join(destination_path, file)

        # Move the file to the new destination
        shutil.move(source_file_path, destination_file_path)
        print(f"Moved: {file} to {destination_file_path}")
```
